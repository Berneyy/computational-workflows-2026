### What is differential abundance analysis?

Differential abundance analysis is a statistical method, that compares e.g., gene expression between different conditions/groups etc. of samples. With those statistics, one can identify possible abundance difference between those oberseved groups/conditions. In medical DAA, this might help to get a deeper look into what genes are targeted by treatments/drugs.

Give the most important plots from the report:

![PCA_plots](./test/results_1/plots/exploratory/rnaseq_deseq2_gsea/treatment/png/pca2d.png)

The PCA depicts the six samples of the two treatments (mCherry and hND6) in two dimensional space. This is done via a dimensionality reduction into the most principle components and has the goal to depict highly dimensional data in a comprehensible way. Furthermore, it gives an impression of how similar samples are, according to their position in the 2D space. If they lie close together, they are very similar. In our case, if points are close together, the should have similar gene expression.
In this case, the samples group for each method, but are still very close to each other. Hence, they don't form clear, distinct clusters. Meaning, according to this PCA, most samples don't seem to have highly differing gene expression, except for SRX8042385.


![expression_heatmap](./test/results_1/report/gsea/rnaseq_deseq2_gsea/treatment_mCherry_hND6__SRP254919/mh.all.v2022.1.Mm.symbols/treatment_mCherry_hND6__SRP254919.mh.all.v2022.1.Mm.symbols.heat_map_1.png)

This heatmap illustrates the gen expression of a selection of genes (on the y-axis) for all six samples of the two treatments (x-axis). These selected genes might be the most variable genes, or genes picked via other criteria. Unfortunately, I'm not entirely sure.
Nevertheless, the heatmap illustrates very clearly different gene expression between both treatments for the shown genes. In roughly the upper half of shown genes, the hND6 treated samples show high abundance overall, whereas the genes are lowly abundant in the mCherry treated samples. In the lower half, it is the other way around. The mCherry treated samples mostly show high abundance of these genes, whereas the hND6 treated samples show low abundance overall.
This depicts very clearly a difference in gene expression for the given genes between both treatments.
